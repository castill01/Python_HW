```python
#  Sort and display the top five schools in overall passing rate
# 
# Each row is a different school.
# The 'reading_score' column shows the average reading score for that school.
# The 'math_score' column shows the average math score for that school.
# This operation effectively summarizes the performance of each school by calculating the average scores
#, allowing for easy comparison between schools.


# Calculate average score for each school
school_avg = school_data.groupby('school_name').agg({
    'reading_score': 'mean',
    'math_score': 'mean'
})
#print(school_avg)


#this was so simple i couldn't  figure out on my own
# Calculate overall average score by creating a new column
school_avg['avg_score'] = (school_avg['reading_score'] + school_avg['math_score']) / 2
#print(school_avg)

# Sort schools by average score in descending order and get top 5
top_5_schools = school_avg.sort_values('avg_score', ascending=False).head()

# Round the numeric columns to 2 decimal places
columns_to_round = ['reading_score', 'math_score', 'avg_score']
top_5_schools[columns_to_round] = top_5_schools[columns_to_round].round(2)

# Display the results
print(top_5_schools[['reading_score', 'math_score', 'avg_score']])
```

```python
# Calculate total budget for each school
#Note there are multiple rows with the same dollar for each school. 
school_budgets = school_data.groupby('school_name')['budget'].first().reset_index()

# Sort schools by budget in descending order
school_budgets = school_budgets.sort_values('budget', ascending=False)

# Format budget as currency
school_budgets['budget'] = school_budgets['budget'].apply(lambda x: f"${x:,.2f}")

# Display the results
print("Total Budget per School:")
print(school_budgets)
```

```python
# Calculate per student budget

school_budgets = school_data.groupby('school_name').agg({
    'budget': 'first',
    'Student ID': 'nunique'
}).reset_index()

school_budgets['Per Student Budget'] = school_budgets['budget'] / school_budgets['Student ID']

# Sort schools by per-student budget in descending order
school_budgets = school_budgets.sort_values('Per Student Budget', ascending=False)

# Format per-student budget as currency
school_budgets['Per Student Budget'] = school_budgets['Per Student Budget'].apply(lambda x: f"${x:.2f}")

# Display the results
print("Per Student Budget by School:")
print(school_budgets[['school_name', 'Per Student Budget']])

```

```python
# Cacluate the avg math and reading score

# Group by school and calculate average math and reading scores
school_averages = school_data.groupby('school_name').agg({
    'math_score': 'mean',
    'reading_score': 'mean'
}).reset_index()

# Round the scores to two decimal places
school_averages['math_score'] = school_averages['math_score'].round(2)
#print(school_averages)
school_averages['reading_score'] = school_averages['reading_score'].round(2)
#print(school_averages)

# Rename columns for clarity
school_averages = school_averages.rename(columns={
    'math_score': 'Avg Math Score',
    'reading_score': 'Avg Reading Score'
})

# Sort schools alphabetically
school_averages = school_averages.sort_values('school_name')

# Print the results
print("Average Math and Reading Scores per School:")
print(school_averages)

```
